```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df=pd.read_csv('movies_dataset.csv')
df.head(3)
df.info()
df.duplicated().sum()
df.isna().sum()
```

```python
df['Country'].value_counts()
```

```python
numrcal_col=['BudgetUSD', 'US_BoxOfficeUSD', 'Global_BoxOfficeUSD',
       'Opening_Day_SalesUSD', 'One_Week_SalesUSD', 'IMDbRating', 'RottenTomatoesScore', 'NumVotesIMDb', 'NumVotesRT']
```

```python
for clo in numrcal_col:
    sns.histplot(df[clo],bins=50, kde=True)
    plt.title(f"Distrib of {clo}")
    plt.show()
```

```python
df.columns
```

```python
catogre_col=['Genre','Country','Director','LeadActor']
```

```python
for col in catogre_col:
    df[col].value_counts().head(20).plot(kind='bar')
    plt.title(f"top 20 {col} counts ")
    plt.show()
```

```python
df.columns
```

```python
sns.scatterplot(data=df ,x='BudgetUSD',y='Global_BoxOfficeUSD',alpha=0.3)
plt.title('BudgetUSD vs Global Box ')
plt.xscale('log')
plt.yscale('log')
plt.show()
```

```python
sns.scatterplot(data= df, x='IMDbRating',y='RottenTomatoesScore',alpha=0.3)
plt.title('IMDbRating  vs RottenTomatoesScore ')
plt.show()
```

```python
df.columns
```

```python
Genre_year_stats=df.groupby(['Genre', 'ReleaseYear']).agg({
     'Global_BoxOfficeUSD':'mean',
     'BudgetUSD':'mean',
     'IMDbRating':'mean',
     'RottenTomatoesScore':'mean'


}).reset_index()
```

```python
Genre_year_stats
```

```python

Genre_year_stats.columns
```

```python
plt.figure(figsize=(15,10))
sns.lineplot(data=Genre_year_stats ,x='ReleaseYear',y='Global_BoxOfficeUSD',hue='Genre')
plt.title("Average Glopal office py Genre over years ")
plt.show()
```

```python
plt.figure(figsize=(15,10))
sns.lineplot(data=Genre_year_stats.query("Genre == 'Action'") ,x='ReleaseYear',y='Global_BoxOfficeUSD',hue='Genre')
plt.title("Average Glopal office py Genre over years ")
plt.show()
```

```python
top_Actors=df.groupby('LeadActor')['IMDbRating'].mean().sort_values(ascending=False).head(20)
```
